# Part 1 — NumPy Essentials (30 min)

NumPy gives us the `ndarray`: a fast, fixed-type, multi-dimensional array that supports vectorized math.

## 1.1 Array creation

````python
a = np.array([1, 2, 3, 4, 5])
zeros = np.zeros((2, 3))
ones = np.ones((3, 3))
ranged = np.arange(0, 10, 2)       # start, stop, step
lin = np.linspace(0, 1, 5)         # start, stop, num_points

print("array:", a)
print("zeros:\n", zeros)
print("ones:\n", ones)
print("arange:", ranged)
print("linspace:", lin)
````

## 1.2 Indexing and slicing (1D and 2D)

````python
arr1d = np.arange(10)
print(arr1d[3])          # single element
print(arr1d[2:7])        # slice
print(arr1d[::-1])       # reversed

arr2d = np.arange(1, 13).reshape(3, 4)
print(arr2d)
print(arr2d[1, 2])       # row 1, col 2
print(arr2d[:, 1])       # entire column 1
print(arr2d[0:2, 1:3])   # sub-matrix
````

## 1.3 Element-wise arithmetic

````python
x = np.array([1, 2, 3, 4])
y = np.array([10, 20, 30, 40])

print(x + y)
print(x * y)
print(y / x)
print(x ** 2)
print(x + 100)   # broadcasting: scalar applied to every element
````

## 1.4 Aggregation across axes

````python
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print("sum all:", np.sum(matrix))
print("sum axis=0 (down columns):", np.sum(matrix, axis=0))
print("sum axis=1 (across rows):", np.sum(matrix, axis=1))
print("mean:", np.mean(matrix), "| std:", np.std(matrix))
````

## 1.5 Why NumPy matters: vectorized vs. plain-Python loop

````python
n = 2_000_000
py_list = list(range(n))
np_arr = np.arange(n)

start = time.time()
py_result = [v * 2 for v in py_list]
py_time = time.time() - start

start = time.time()
np_result = np_arr * 2
np_time = time.time() - start

print(f"Python loop: {py_time:.4f}s")
print(f"NumPy vectorized: {np_time:.4f}s")
print(f"NumPy was ~{py_time / np_time:.0f}x faster")
````

# Part 3 — Plotting with Matplotlib and Seaborn (40 min)

## 3.1 Matplotlib basics

````python
plt.figure(figsize=(6, 4))
plt.plot(df.sort_values("years_experience")["years_experience"].values[:50])
plt.xlabel("Index")
plt.ylabel("Years of experience")
plt.title("Line plot example")
plt.show()
````

## 3.2 Histogram — single-variable distribution

````python
plt.figure(figsize=(6, 4))
plt.hist(df["salary"].dropna(), bins=20)
plt.xlabel("Salary")
plt.ylabel("Count")
plt.title("Distribution of Salary")
plt.show()
````

## 3.3 Scatter plot — two-variable relationship

````python
plt.figure(figsize=(6, 4))
plt.scatter(df["years_experience"], df["salary"])
plt.xlabel("Years of experience")
plt.ylabel("Salary")
plt.title("Experience vs. Salary")
plt.show()
````

## 3.4 Box plot — comparing distributions across categories

````python
plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x="department", y="salary")
plt.title("Salary by Department")
plt.show()
````

## 3.5 Heatmap — visualizing a numeric matrix

````python
numeric_cols = ["age", "years_experience", "salary", "performance_score", "satisfaction_score"]
corr = df[numeric_cols].corr()

plt.figure(figsize=(6, 5))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
````

## 3.6 Subplots — a grid of multiple plots

````python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].hist(df["performance_score"].dropna(), bins=15, color="steelblue")
axes[0].set_title("Performance Score Distribution")
axes[0].set_xlabel("Performance score")
axes[0].set_ylabel("Count")

axes[1].scatter(df["satisfaction_score"], df["performance_score"], color="darkorange")
axes[1].set_title("Satisfaction vs. Performance")
axes[1].set_xlabel("Satisfaction score")
axes[1].set_ylabel("Performance score")

plt.tight_layout()
plt.show()
````
